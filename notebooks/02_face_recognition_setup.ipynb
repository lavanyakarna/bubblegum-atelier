# Face Recognition Setup (LBPH)

The API already exposes `/register-face` and `/recognize-face` (see `app/routers/vision.py`).
This notebook is only for testing the pipeline locally / offline with sample images before wiring it through the API.

Use consenting sample faces (e.g. an LFW subset) or your own webcam photos — never scrape real customer photos without consent.

```python
import sys, os
sys.path.insert(0, '..')
import cv2
from app.services.cv_service import FaceRecognitionService, detect_faces, to_grayscale, resize_image

face_service = FaceRecognitionService()
```

```python
# Register a customer: point this at 3-5 photos of one person
sample_paths = ['../data/sample_faces/alice_1.jpg', '../data/sample_faces/alice_2.jpg']

crops = []
for p in sample_paths:
    img = cv2.imread(p)
    faces = detect_faces(img)
    if len(faces) == 0:
        print('no face found in', p)
        continue
    x, y, w, h = faces[0]
    crop = resize_image(to_grayscale(img)[y:y+h, x:x+w], 200, 200)
    crops.append(crop)

print(face_service.register_customer('Alice', crops))
```

```python
# Test recognition on a new, unseen photo of the same person
test_img = cv2.imread('../data/sample_faces/alice_3.jpg')
faces = detect_faces(test_img)
x, y, w, h = faces[0]
crop = resize_image(to_grayscale(test_img)[y:y+h, x:x+w], 200, 200)
print(face_service.recognize(crop))
```
